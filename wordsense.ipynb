```python
#import
import nltk
from nltk.corpus import senseval
from nltk.classify import NaiveBayesClassifier, accuracy
from nltk.tokenize import word_tokenize
import random

nltk.download('senseval')
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
```

```python
#feature extraction
def get_features(instance):
    features = {}
    for i, token in enumerate(instance.context):
        word = token[0].lower()   # index 0 = word, index 1 = pos, index 2 = position
        features[f'word_{i}'] = word
    return features

print("Feature extractor defined.")
```

```python
#load senseval
instances = senseval.instances('hard.pos')

print(f"Total instances loaded : {len(instances)}")
print(f"Sample instance senses : {set(inst.senses for inst in instances[:20])}")
```

```python
#prepare labelled set
labelled = [(get_features(inst), inst.senses[0]) for inst in instances]


random.seed(42)
random.shuffle(labelled)

split = int(0.8 * len(labelled))
train_set = labelled[:split]
test_set  = labelled[split:]

print(f"Training samples : {len(train_set)}")
print(f"Testing  samples : {len(test_set)}")
```

```python
#train Naïve Bayes classifier
classifier = NaiveBayesClassifier.train(train_set)
print("Naïve Bayes classifier trained successfully.")
```

```python
#aacuracy
acc = accuracy(classifier, test_set)
print(f"(a) Accuracy : {acc * 100:.2f}%")
```

```python
#most informative features
print("(b) Most Informative Features:")
classifier.show_most_informative_features(10)
```

```python
#predictive sense
sentence = "He is a hard worker"
tokens   = word_tokenize(sentence.lower())

# Build a feature dict identical to the training features
test_features = {f'word_{i}': w for i, w in enumerate(tokens)}

predicted_sense = classifier.classify(test_features)
print(f"(c) Predicted Sense for '{sentence}' : {predicted_sense}")
```
