```python
#Q1 of exp list
#1. tokenization and stop word removal
# 1. Import required libraries
import nltk
import re
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords

# 2. Input your text
text = input("Enter the Sentence:")#Natural Language Processing is fun. Tokenization is the first step!"

# 3. Sentence tokenization
sentences = sent_tokenize(text)
print("Sentences:")
print(sentences)

# 3. Word tokenization
tokens = word_tokenize(text)
print("\n All Tokens:")
print(tokens)

# 4 & 5. Filter unwanted tokens and ensure valid English alphabet tokens
filtered_tokens = [
    token.lower()
    for token in tokens
    if re.fullmatch(r"[A-Za-z]+", token)
]

print("\nFiltered Tokens (English alphabets only):")
print(filtered_tokens)

# Remove stopwords
stop_words = set(stopwords.words('english'))

final_tokens = [
    token for token in filtered_tokens
    if token not in stop_words
]

print("\nTokens after Stopword Removal:")
print(final_tokens)
```

```python
#Q2 Stemming 
import nltk
import re
import string
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download necessary NLTK data (will only download if not already present)
#nltk.download('punkt', quiet=True)
#nltk.download('stopwords', quiet=True)
#nltk.download('punkt_tab', quiet=True) # Ensure punkt_tab is downloaded for word_tokenize
#print("NLTK resources loaded successfully.")

# Instantiate Porter Stemmer
porter_stemmer = PorterStemmer()
#print("PorterStemmer instantiated.")

def preprocess_text(text):
    # 1. Convert to lowercase
    text_lower = text.lower()

    # 2. Remove punctuation
    translator = str.maketrans('', '', string.punctuation)
    text_no_punct = text_lower.translate(translator)

    # 3. Tokenize the cleaned text
    tokens = word_tokenize(text_no_punct)

    # 4. Get English stop words for efficient lookup
    stop_words = set(stopwords.words('english'))

    # 5. Filter and stem tokens
    normalized_tokens = []
    for token in tokens:
        if token.isalpha() and token not in stop_words:
            stemmed_token = porter_stemmer.stem(token)
            normalized_tokens.append(stemmed_token)

    return normalized_tokens

# --- Get user input and apply preprocessing ---
user_input = input("Enter the text you want to normalize: ")

# Apply the preprocessing function
processed_output = preprocess_text(user_input)

print("--- Normalization Results ---")
print(f"\n Original Text: {user_input}")
print(f"\n Tokens: {tokens}")
print(f"\n Normalized Tokens after stemming: {processed_output}")
```

```python
#POS tag
import nltk
from nltk.tokenize import word_tokenize

text = input("Enter sentence: ")

tokens = word_tokenize(text)

def rule_based_tagger(words):

    tags = []

    for word in words:

        w = word.lower()

        # Punctuation
        if word in ['.', ',', '!', '?', ';', ':']:
            tag = 'Punctuation'

        # Pronouns
        elif w in ['i', 'you', 'he', 'she', 'we', 'they', 'me', 'my']:
            tag = 'Pronoun'

        # Determiners
        elif w in ['the', 'a', 'an']:
            tag = 'Determiner'

        # Common verbs
        elif w in ['is', 'am', 'are', 'was', 'were', 'help', 'reset', 'access']:
            tag = 'Verb'

        # Adverbs
        elif w.endswith("ly"):
            tag = 'Adverb'

        # Adjectives
        elif w in ['quick', 'lazy', 'unable']:
            tag = 'Adjective'

        # Nouns
        elif w in ['team', 'banking', 'account',
                   'password', 'yesterday']:
            tag = 'Noun'

        # Proper nouns
        elif word[0].isupper() and w not in ['i', 'please', 'hello','good morning']:
            tag = 'Proper Noun'

        else:
            tag = 'Unknown'

        tags.append((word, tag))

    return tags

rule_tags = rule_based_tagger(tokens)

print("\nRule-Based POS Tags:\n")

for word, tag in rule_tags:
    print(f"{word} → {tag}")
```

```python
#Q2 LEMMETIZATION
import nltk
import string
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download resources
for pkg in ['punkt', 'stopwords', 'wordnet', 'omw-1.4']:
    nltk.download(pkg, quiet=True)

lemmatizer = WordNetLemmatizer()

def preprocess_text(text):
    text = text.lower()
    text = text.translate(str.maketrans('', '', string.punctuation))

    tokens = word_tokenize(text)
    stop_words = set(stopwords.words('english'))

    normalized_tokens = []
    for token in tokens:
        if token.isalpha() and token not in stop_words:
            # Try verb first, then noun
            lemma = lemmatizer.lemmatize(token, 'v')
            lemma = lemmatizer.lemmatize(lemma, 'n')
            normalized_tokens.append(lemma)

    return normalized_tokens

# Test
text = input("Enter the Sentence: ")#"Students are studying advanced machine learning algorithms."
print(preprocess_text(text))
```

```python
#Q5 1. tokenization, 2. lemmatization 
#3. edit distance
# Spell Correction using Levenshtein Edit Distance

import nltk
from nltk.corpus import words

nltk.download('words')
def edit_distance(a, b):
    m, n = len(a), len(b)

    # Create matrix
    dp = [[0 for j in range(n + 1)] for i in range(m + 1)]

    # Fill first row and column
    for i in range(m + 1):
        dp[i][0] = i

    for j in range(n + 1):
        dp[0][j] = j

    # Fill matrix
    for i in range(1, m + 1):
        for j in range(1, n + 1):

            if a[i - 1] == b[j - 1]:
                cost = 0
            else:
                cost = 1

            dp[i][j] = min(
                dp[i - 1][j] + 1,        # Deletion
                dp[i][j - 1] + 1,        # Insertion
                dp[i - 1][j - 1] + cost # Substitution
            )

    return dp[m][n], dp


# Existing Word Corpus
corpus = words.words() #corpus = ["spelling", "language", "receive", "running"]

# Input Word
word = input("Enter misspelled word: ").lower()

# Find closest word
best_word = ""
min_dist = 999

for w in corpus:

    dist, matrix = edit_distance(word, w)

    if dist < min_dist:
        min_dist = dist
        best_word = w
        best_matrix = matrix


# Output
print("\nSuggested Word:", best_word)
print("Edit Distance:", min_dist)

print("\nEdit Distance Matrix:\n")

for row in best_matrix:
    print(row)
```

```python
#Q3 morphological analysis

import spacy

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

def get_features(token):
    pos = token.pos_
    morph = token.morph
    feature = ""

    if pos == "NOUN":
        number = morph.get("Number")
        if number:
            if number[0] == "Plur":
                feature = "noun, plural"
            else:
                feature = "noun, singular"
        else:
            feature = "noun"

    elif pos == "VERB":
        tense = morph.get("Tense")
        aspect = morph.get("Aspect")

        if tense and tense[0] == "Past":
            feature = "verb, past tense"

        elif tense and tense[0] == "Pres":
            if aspect and aspect[0] == "Prog":
                feature = "verb, present continuous"
            else:
                feature = "verb, present"
        else:
            feature = "verb"

    return feature


# Take input
text = input("Enter a sentence: ").lower()

# Process text
doc = nlp(text)

print("\nOutput:")
for token in doc:
    if token.is_alpha and token.pos_ in ["NOUN", "VERB"]:
        root = token.lemma_
        features = get_features(token)
        print(f"{token.text} → {root} → {features}")
```

```python
#Q3 word generation
def generate_word(root, feature):
    root, feature = root.lower(), feature.lower()

    vowels = "aeiou"
    irr = {
        "plural": {"child": "children", "man": "men", "woman": "women", "foot": "feet", "tooth": "teeth"},
        "past tense": {"go": "went", "eat": "ate", "run": "ran", "is":"was", "do":"did"},
        "comparative": {"good": "better", "bad": "worse"}
    }

    if root in irr.get(feature, {}):
        return irr[feature][root]

    if feature == "plural":
        return (root + "es" if root.endswith(("s","x","z","ch","sh"))
                else root[:-1] + "ies" if root.endswith("y") and root[-2] not in vowels
                else root + "s")

    if feature == "past tense":
        return (root + "d" if root.endswith("e")
                else root[:-1] + "ied" if root.endswith("y") and root[-2] not in vowels
                else root + root[-1] + "ed" if len(root)>=3 and root[-1] not in vowels and root[-2] in vowels and root[-3] not in vowels
                else "Feature not Supported")

    if feature == "comparative":
        return (root[:-1] + "ier" if root.endswith("y")
                else root + root[-1] + "er" if len(root)>=3 and root[-1] not in vowels and root[-2] in vowels and root[-3] not in vowels
                else root + "er")

    return "Feature not supported."


# input/output
print(generate_word(input("Word: "), input("Feature(plural/past tense/ comparative: ")))
```

```python
#Q7 tokenization, stemming, N gram  Bigram Language Model - Sentence Probability & Next Word Prediction

from collections import defaultdict
from nltk.util import bigrams
from nltk.tokenize import word_tokenize
import nltk

nltk.download('punkt')

# Training Corpus
corpus = [
    "I like AI",
    "I like machine learning",
    "I like AI"
]

# Dictionaries
unigram_count = defaultdict(int)
bigram_count = defaultdict(int)

# Training
for sentence in corpus:
    tokens = ['<s>'] + word_tokenize(sentence.lower()) + ['</s>']

    # Unigram Count
    for word in tokens:
        unigram_count[word] += 1

    # Bigram Count
    for bg in bigrams(tokens):
        bigram_count[bg] += 1

# Display Unigrams
print("Unigrams:\n")
for word, count in unigram_count.items():
    print(word, ":", count)

# Display Bigrams
print("\nBigrams:\n")
for bg, count in bigram_count.items():
    print(bg, ":", count)

# Sentence Probability
def sentence_probability(sentence):
    tokens = ['<s>'] + word_tokenize(sentence.lower()) + ['</s>']
    probability = 1.0

    for bg in bigrams(tokens):
        w1, w2 = bg

        prob = bigram_count[bg] / unigram_count[w1]

        probability *= prob

        print(f"P({w2}|{w1}) = {prob}")

    return probability

# Next Word Prediction
def predict_next_word(word):
    candidates = {}

    for (w1, w2), count in bigram_count.items():

        if w1 == word:
            candidates[w2] = count / unigram_count[w1]

    if candidates:
        return max(candidates, key=candidates.get)
    else:
        return "No prediction"

# Input Sentence
sentence = input("Enter the sentence") #"I like NLP"

# Probability
prob = sentence_probability(sentence)

print("\nSentence Probability:", round(prob,2))

# Predict Next Word
word = "like"

print("\nPredicted Next Word after", word, ":", predict_next_word(word))
```
