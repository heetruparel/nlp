```python
#importing
import nltk

nltk.download('punkt_tab')
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('omw-1.4')
```

```python
#Tokenization + Stopword Removal
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

text = "This product quality is very good and affordable"

tokens = word_tokenize(text.lower())

filtered = [w for w in tokens if w not in stopwords.words('english')]

print(filtered)
```

```python
#Filtration
import re

text = "Amazing product!!! Visit www.shop.com"

clean = re.sub(r'http\S+|[^a-zA-Z ]', '', text)

print(clean)
```

```python
#unnigram,bigram,trigram
from nltk.util import ngrams

text = "I love NLP"

words = text.split()

print(list(ngrams(words,1)))
print(list(ngrams(words,2)))
print(list(ngrams(words,3)))
```
